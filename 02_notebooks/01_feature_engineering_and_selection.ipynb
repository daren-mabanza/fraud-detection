# Note explicative du Notebook

Ce notebook met en place la première étape du projet : préparer les données de transactions avant l’analyse exploratoire et la modélisation.

### 1. Import et remise en forme

Les données sont chargées depuis un fichier parquet puis certains types sont corrigés (dates, identifiants en `object`, etc.).  
Les variables sont ensuite renommées pour obtenir des noms clairs et orientés métier (ex. `amt` → `montant_transaction`, `is_fraud` → `target`).

### 2. Nettoyage

- Nettoyage du champ `nom_magasin`.
- Suppression des variables non exploitables ou sensibles :
  - identifiants directs (`numero_carte`, `prenom`, `nom`)
  - variables posant des enjeux éthiques (`sexe`)
  - informations redondantes ou trop fines (`adresse_client`, `code_postal_client`, `timestamp_unix_transacation`, etc.)

L’objectif est de conserver uniquement des variables pertinentes pour la détection de fraude.

### 3. Création de variables (feature engineering)

Ajout de variables dérivées pour enrichir l’information :

- **Variables temporelles** : année, mois, jour, heure de transaction  
- **Distance domicile–magasin** : calculée à partir des coordonnées GPS  
- **Âge du client** : dérivé de la date de naissance

Ces variables visent à capturer :
- des comportements inhabituels dans le temps (heures atypiques, saisonnalité),
- des incohérences géographiques,
- des effets liés au profil du client.

### 4. Contrôle qualité et export

Un contrôle qualité est réalisé avec Great Expectations :
- vérification du nombre de colonnes,
- présence de la variable cible,
- absence de valeurs nulles sur `target`.

Si les conditions sont validées, la table nettoyée est exportée en parquet (`data_fraud_processed.parquet`) pour l’EDA et la suite du pipeline

---

En résumé, ce notebook transforme les données brutes issues du dataset Kaggle en une table propre, enrichie et prête pour l’analyse exploratoire et la modélisation.
