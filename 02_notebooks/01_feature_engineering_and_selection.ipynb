# Note explicative du Notebook

Ce notebook a pour objectif de préparer les données brutes avant les phases d’analyse exploratoire et de modélisation.  
Il regroupe l’import des données, le nettoyage, la création de variables ainsi que l’export d’une table prête à être utilisée.

---

## Import et fusion des données

Les fichiers `fraudTrain.csv` et `fraudTest.csv` sont chargés puis regroupés en une seule table.

Plusieurs opérations sont réalisées lors de cette étape :
- conversion des types de variables (dates en datetime, identifiants en object)
- suppression de la colonne technique `Unnamed: 0`
- concaténation des deux jeux de données

Ce regroupement permet de garder la main sur le découpage train, validation et test, en particulier dans une logique temporelle cohérente avec le cas d’usage.

---

## Renommage des variables

Les colonnes sont renommées avec des intitulés plus explicites et orientés métier.

Par exemple :
- `amt` devient `montant_transaction`
- `trans_date_trans_time` devient `date_heure_transaction`
- `is_fraud` devient `target`

L’objectif est d’améliorer la lisibilité du dataset et de faciliter l’interprétation des résultats par la suite.

---

## Nettoyage ciblé

Un nettoyage est appliqué sur certaines variables catégorielles, notamment `nom_magasin`, afin de supprimer des préfixes techniques inutiles.

Cela permet d’éviter une multiplication artificielle des modalités et de garder une représentation plus propre des données.

---

## Création de variables

Plusieurs variables sont construites à partir des données existantes afin d’enrichir l’information disponible pour la modélisation.

### Variables temporelles

À partir de `date_heure_transaction`, les variables suivantes sont créées :
- `mois_transaction`
- `jour_transaction`
- `heure_transaction`

Elles permettent de capturer des effets liés au moment de la transaction.

### Variable géographique

- `distance_domicile_magasin`

Elle est calculée à partir des coordonnées du client et du commerçant, puis exprimée en kilomètres.  
Les coordonnées brutes ne sont pas conservées, la distance étant plus directement exploitable.

### Variable démographique

- `age_client`

Elle est calculée à partir de la date de transaction et de la date de naissance.

---

## Suppression de variables

Certaines variables sont supprimées car non exploitables, redondantes ou non pertinentes dans le cadre du modèle :

- identifiants uniques (`numero_carte`, `trans_num`)
- informations personnelles (`prenom`, `nom`, adresse complète)
- variable sensible (`sexe`)
- variables redondantes (`timestamp_unix_transaction`, coordonnées brutes)
- variables remplacées par des variables dérivées (`date_heure_transaction`, `date_naissance_client`)

Cette étape permet de réduire le bruit, de simplifier le dataset et de limiter les risques de fuite d’information.

---

## Contrôle qualité

Un contrôle est réalisé avec Great Expectations afin de vérifier la cohérence des données :

- présence de la variable cible `target`
- absence de valeurs manquantes sur cette variable
- conformité du nombre de colonnes

L’export n’est effectué que si ces conditions sont respectées.

---

## Export

La table finale, nettoyée et enrichie, est exportée au format parquet.  
Elle servira de base pour le notebook suivant consacré à l’analyse exploratoire des données.
