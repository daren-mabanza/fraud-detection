# Note explicative du Notebook

Ce notebook prépare les données brutes avant l’analyse exploratoire et la modélisation.  
Il regroupe l’import, le nettoyage, la création de variables et l’export d’une table prête à être exploitée.

---

### 1. Import et fusion des données

Les fichiers `fraudTrain.csv` et `fraudTest.csv` sont :

- importés
- typés correctement (dates en datetime, identifiants en object)
- débarrassés de la colonne technique `Unnamed: 0`
- concaténés en une seule table

Cette fusion permet ensuite de réaliser un découpage temporel cohérent entre entraînement, validation et test.

---

### 2. Renommage des variables

L’ensemble des colonnes est renommé avec des intitulés explicites et orientés métier.  

Exemples :
- `amt` devient `montant_transaction`
- `trans_date_trans_time` devient `date_heure_transaction`
- `is_fraud` devient `target`

Ce renommage améliore la lisibilité et facilite l’interprétation des résultats.

---

### 3. Nettoyage ciblé

La variable `nom_magasin` est modifiée afin de supprimer un préfixe technique présent dans les valeurs.  
Cela permet d’éviter une hétérogénéité inutile dans les modalités.

---

### 4. Création de variables

Plusieurs variables sont construites à partir des données existantes.

#### Variables temporelles

À partir de `date_heure_transaction` :

- `annee_transaction`
- `mois_transaction`
- `jour_transaction`
- `heure_transaction`

Ces variables permettent de capter d’éventuels effets saisonniers ou horaires.

#### Variable géographique

- `distance_domicile_magasin`

Elle est calculée à partir des coordonnées du client et du commerçant, puis exprimée en kilomètres.  
Les coordonnées brutes ne sont pas conservées, la distance synthétisant l’information pertinente.

#### Variable démographique

- `age_client`

Calculé à partir de l’année de transaction et de la date de naissance.

---

### 5. Suppression de variables

Certaines colonnes sont supprimées car non exploitables ou non pertinentes pour la modélisation :

- identifiants uniques (`numero_carte`)
- informations personnelles (`prenom`, `nom`)
- variable sensible (`sexe`)
- informations redondantes (`timestamp_unix_transacation`, coordonnées brutes)
- variables remplacées par des features dérivées (`date_heure_transaction`, `date_naissance_client`)

Cette étape permet de limiter le bruit et les risques de fuite d’information.

---

### 6. Contrôle qualité

Un contrôle est réalisé avec Great Expectations :

- vérification de la présence de la variable `target`
- vérification de l’absence de valeurs nulles sur `target`
- vérification du nombre total de colonnes attendu

L’export n’est effectué que si l’ensemble des conditions est validé.

---

### 7. Export

La table nettoyée et enrichie est exportée au format parquet.  
Elle constitue la base du notebook suivant consacré à l’analyse exploratoire.
